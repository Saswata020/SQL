Question 15 (Very Important Concept)

Write a query to:

Show employee name and salary
Only for employees who earn the HIGHEST salary in the company

##scalar subqueries (returning one value).

````python
SELECT E.NAME AS EMP_NAME,
       E.SALARY
FROM EMPLOYEES E
WHERE SALARY = (
    SELECT MAX(SALARY)
    FROM EMPLOYEES
);
````

## Correlated subqueries.

Show employee name, salary, and department_id
Only for employees who earn the HIGHEST salary within their own department

````python
SELECT E.name,
       E.salary,
       E.department_id
FROM employees E
WHERE E.salary = (
    SELECT MAX(E2.salary)
    FROM employees E2
    WHERE E2.department_id = E.department_id
);
````

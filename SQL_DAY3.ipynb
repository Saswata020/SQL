Question 7

Write a query to:

Show employee name, salary, and department_name

Only show employees earning more than 50,000

Sort by salary descending.

````python
SELECT E.name AS employee_name,
       E.salary,
       D.department_name
FROM employees E
INNER JOIN departments D
    ON E.department_id = D.department_id
WHERE E.salary > 50000
ORDER BY E.salary DESC;
````

Question 10

Write a query to:

Show all employees (even those without departments)
Display name, salary, and department_name
Sort by salary descending

````python
SELECT E.name AS employee_name,
       E.salary,
       D.department_name
FROM employees E
LEFT JOIN departments D
    ON E.department_id = D.department_id
ORDER BY E.salary DESC;
````

Question 11 (Tricky and common in interviews)

Using the same LEFT JOIN structure:

Write a query to:

Show employees who do NOT have a department assigned.

````python
SELECT E.name AS employee_name,
       E.salary,
       D.department_name
FROM employees E
LEFT JOIN departments D
    ON E.department_id = D.department_id
WHERE D.department_id IS NULL
ORDER BY E.salary DESC;
````

Question 12 (Real-world SQL)

Write a query to:

Show department_name and number of employees in each department
Include departments with ZERO employees

````python
SELECT D.department_name,
       COUNT(E.id) AS employee_count
FROM departments D
LEFT JOIN employees E
    ON E.department_id = D.department_id
GROUP BY D.department_name
ORDER BY employee_count DESC;
````

Question 13 (Harder – Interview Quality)

Write a query to:

Show department_name and employee_count
But only show departments that have MORE THAN 3 employees
Still include departments with zero employees in logic (but they won’t pass the filter)

````python
SELECT D.department_name,
       COUNT(E.id) AS employee_count
FROM departments D
LEFT JOIN employees E
    ON E.department_id = D.department_id
GROUP BY D.department_name
HAVING COUNT(E.id)>3
ORDER BY employee_count DESC;
````

Question 14 (Interview Favorite)

Write a query to:

Show employee name and salary
Only for employees whose salary is ABOVE the average salary of ALL employees

Use a subquery.

````python
SELECT name,
       salary
FROM employees
WHERE salary > (
    SELECT AVG(salary)
    FROM employees
);
````

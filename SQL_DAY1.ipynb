Question 1

Write a SQL query to:

Show name and marks of students who scored more than 50

```
SELECT name, marks
FROM students
WHERE marks > 50;


Question 2

Write a query to:

Show all students sorted by marks from highest to lowest

```
SELECT *
FROM students
ORDER BY marks DESC;


Question 3

Write a query to:

Show name and marks of students who scored 50 or more, sorted by marks ascending

```
SELECT name, marks
FROM students
WHERE marks >= 50
ORDER BY marks ASC;


```
Question 4

Table: employees

id	name	department	salary

Write a query to:

Show name and salary of employees who work in IT and earn more than 50,000, sorted by salary descending

```
SELECT name, salary
FROM employees
WHERE department IN ('IT')
  AND salary > 50000
ORDER BY salary DESC;


Question 5

Write a query to:

Show department and average salary for each department

```
SELECT department, AVG(salary) AS average_salary
FROM employees
GROUP BY department;
